```python
import matplotlib.pyplot as py
import numpy as np
xpoints=([2,4])
ypoints=([3,6])
py.subplot(2,3,1)
py.title("title 1")
py.xlabel("xlabel",color="blue")
py.ylabel("ylabel",color="blue")
py.plot(xpoints,ypoints,'-',marker='o')

xpoints=([2,5,2,5,2])
ypoints=([4,3,4,3,4])
py.subplot(2,3,2)
py.plot(xpoints,marker='*',color='orange',mfc="black")
py.plot(ypoints,'--')


xpoints=np.array([2,4,3,6])
py.subplot(2,3,3)
py.plot(xpoints,marker='o',ms=10,mec='black',mfc='yellow')


xpoints=np.array([2,4,6,8,10,12,14])
py.subplot(2,3,4)
py.plot(xpoints)
py.grid()


xpoints=np.array([2,6,9,3,8])
ypoints=np.array([1,2,3,4,2])
py.title("title2")
py.subplot(2,3,5)
color=(["red","blue","green","black","hotpink"])
sizes=([200,120,78,150,260])
py.scatter(xpoints,ypoints,c=color,s=sizes,alpha=0.5)


xpoints=np.array([2,4,6,8])
py.subplot(2,3,6)
py.hist(xpoints,width=0.1,color="red")
py.suptitle("Graph")
py.tight_layout()
py.show()
```
