```python
import matplotlib.pyplot as no
import numpy as np
xpoints=np.array([35,20,20,25])
colors=(["yellow","black","maroon","hotpink"])
labels=["35","20","20","25"]
myexplode=([0.1,0.1,0.1,0.1])
no.pie(xpoints,startangle=90,colors=colors,labels=labels,explode=myexplode,shadow=True)
no.legend()
no.show()
```
